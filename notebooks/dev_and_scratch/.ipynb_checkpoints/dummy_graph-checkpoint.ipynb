```python
import networkx as nx
import matplotlib.pyplot as plt

# Dummy graph: 5 nodes, 6 edges (5-cycle + one chord)
edges = [(0, 1), (1, 2), (2, 3), (3, 4), (4, 0), (1, 3)]
n_vertices = 5

G = nx.Graph()
G.add_edges_from(edges)

pos = {
    0: (0.0, 1.0),
    1: (0.95, 0.31),
    2: (0.59, -0.81),
    3: (-0.59, -0.81),
    4: (-0.95, 0.31),
}

fig, ax = plt.subplots(figsize=(7, 7))
nx.draw_networkx_edges(G, pos, ax=ax, width=2.5, edge_color='#444444')
nx.draw_networkx_nodes(G, pos, ax=ax, node_size=1400, node_color='#a8d5ff',
                        edgecolors='#1a1a1a', linewidths=2)
nx.draw_networkx_labels(G, pos, ax=ax, font_size=20, font_weight='bold')

edge_labels = {e: f"e{idx}" for idx, e in enumerate(edges)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, ax=ax,
                               font_size=13, font_color='#b5451b')

ax.set_title("Dummy Graph — 5 nodes, 6 edges (5-cycle + one chord)")
ax.axis('off')
plt.show()
```
